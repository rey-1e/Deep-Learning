```python

# RNN vs LSTM vs GRU
# Reuters Newswire Sequence Classification
# Name: Raj Kanade | Roll No: 26 | PRN: 12413760 | Batch: B3



# 1. INSTALL / IMPORT LIBRARIES

!pip -q install scikit-learn

import tensorflow as tf
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import time
import gc

from tensorflow.keras.datasets import reuters
from tensorflow.keras.preprocessing.sequence import pad_sequences

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, SimpleRNN, LSTM, GRU, Dense, Dropout

from tensorflow.keras.callbacks import EarlyStopping

from sklearn.metrics import (
    accuracy_score,
    precision_recall_fscore_support,
    confusion_matrix,
    classification_report
)


# 2. SET RANDOM SEED

SEED = 42

np.random.seed(SEED)
tf.random.set_seed(SEED)


# 3. PARAMETERS

MAX_FEATURES = 10000
MAX_LEN = 200

EMBEDDING_DIM = 128
HIDDEN_UNITS = 64

EPOCHS = 5
BATCH_SIZE = 128

NUM_CLASSES = 46


# 4. LOAD REUTERS DATASET

print("Loading Reuters dataset...")

(x_train, y_train), (x_test, y_test) = reuters.load_data(
    num_words=MAX_FEATURES
)


# 5. DISPLAY DATASET INFORMATION

print("\n================ DATASET INFORMATION ================")

print("Training samples :", len(x_train))
print("Testing samples  :", len(x_test))
print("Number of classes:", NUM_CLASSES)

print("\nFirst training sample:")
print(x_train[0][:20])

print("\nFirst training label:")
print(y_train[0])


# 6. CHECK SEQUENCE LENGTHS

train_lengths = [len(x) for x in x_train]

print("\n================ SEQUENCE INFORMATION ================")

print("Minimum sequence length :", min(train_lengths))
print("Maximum sequence length :", max(train_lengths))
print("Average sequence length :", round(np.mean(train_lengths), 2))
print("Median sequence length  :", np.median(train_lengths))


# 7. VISUALIZE SEQUENCE LENGTH DISTRIBUTION

plt.figure(figsize=(10, 5))

plt.hist(
    train_lengths,
    bins=50,
    edgecolor='black'
)

plt.axvline(
    MAX_LEN,
    linestyle='--',
    linewidth=2,
    label=f'MAX_LEN = {MAX_LEN}'
)

plt.title("Distribution of Reuters Sequence Lengths")
plt.xlabel("Sequence Length")
plt.ylabel("Number of Articles")
plt.legend()
plt.show()


# 8. VISUALIZE CLASS DISTRIBUTION

class_counts = np.bincount(
    y_train,
    minlength=NUM_CLASSES
)

plt.figure(figsize=(14, 5))

plt.bar(
    range(NUM_CLASSES),
    class_counts
)

plt.title("Reuters Training Dataset - Class Distribution")
plt.xlabel("Class / Topic ID")
plt.ylabel("Number of Articles")

plt.xticks(range(NUM_CLASSES))

plt.show()


# 9. PAD SEQUENCES

print("\nPadding sequences...")

x_train = pad_sequences(
    x_train,
    maxlen=MAX_LEN
)

x_test = pad_sequences(
    x_test,
    maxlen=MAX_LEN
)

print("Training shape:", x_train.shape)
print("Testing shape :", x_test.shape)


# 10. BUILD MODEL

def build_model(kind):
    model = Sequential()

    # mask_zero ignores padding tokens
    model.add(
        Embedding(
            input_dim=MAX_FEATURES,
            output_dim=128,
            input_length=MAX_LEN,
            mask_zero=True
        )
    )

    if kind == "RNN":
        model.add(SimpleRNN(128))
    elif kind == "LSTM":
        model.add(LSTM(128))
    elif kind == "GRU":
        model.add(GRU(128))

    model.add(Dropout(0.3))
    model.add(Dense(64, activation="relu"))
    model.add(Dropout(0.2))
    model.add(Dense(NUM_CLASSES, activation="softmax"))

    model.compile(
        optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
        loss="sparse_categorical_crossentropy",
        metrics=["accuracy"]
    )

    return model

# 11. TRAIN AND EVALUATE MODELS

results = []

histories = {}

predictions = {}

confusion_matrices = {}

training_times = {}

models = {}


for kind in ["RNN", "LSTM", "GRU"]:

    print("\n")
    print("=" * 70)
    print("TRAINING:", kind)
    print("=" * 70)

    # Build model
    model = build_model(kind)

    print("\nModel Summary:")
    model.summary()

    # Early stopping
    early_stopping = EarlyStopping(
        monitor="val_loss",
        patience=2,
        restore_best_weights=True
    )

    # Start timer
    start_time = time.time()

    # Train model
    history = model.fit(
        x_train,
        y_train,
        epochs=EPOCHS,
        batch_size=BATCH_SIZE,
        validation_split=0.2,
        callbacks=[early_stopping],
        verbose=1
    )

    # End timer
    training_time = time.time() - start_time

    training_times[kind] = training_time

    # Store history
    histories[kind] = history.history

    # Prediction

    pred_prob = model.predict(
        x_test,
        verbose=0
    )

    # Convert probabilities to class labels
    pred = np.argmax(
        pred_prob,
        axis=1
    )

    predictions[kind] = pred

    # Evaluation Metrics

    accuracy = accuracy_score(
        y_test,
        pred
    )

    precision, recall, f1, _ = precision_recall_fscore_support(
        y_test,
        pred,
        average="weighted",
        zero_division=0
    )

    # Confusion matrix
    cm = confusion_matrix(
        y_test,
        pred
    )

    confusion_matrices[kind] = cm

    # Store model
    models[kind] = model

    # Number of parameters
    parameters = model.count_params()

    # Store results
    results.append([
        kind,
        accuracy,
        precision,
        recall,
        f1,
        training_time,
        parameters
    ])

    # Print metrics
    print("\n", kind, "RESULTS")

    print("Accuracy :", round(accuracy, 4))
    print("Precision:", round(precision, 4))
    print("Recall   :", round(recall, 4))
    print("F1 Score :", round(f1, 4))

    print("Training Time:",
          round(training_time, 2),
          "seconds")

    print("Parameters:",
          parameters)

    # Classification report
    print("\nClassification Report:")

    print(
        classification_report(
            y_test,
            pred,
            zero_division=0
        )
    )


# 12. CREATE FINAL RESULTS DATAFRAME

df = pd.DataFrame(
    results,
    columns=[
        "Model",
        "Accuracy",
        "Precision",
        "Recall",
        "F1 Score",
        "Training Time (sec)",
        "Parameters"
    ]
)


print("\n")
print("=" * 70)
print("FINAL MODEL COMPARISON")
print("=" * 70)

print(df)


# 13. ROUND RESULTS FOR BETTER DISPLAY

display_df = df.copy()

display_df["Accuracy"] = display_df["Accuracy"].round(4)
display_df["Precision"] = display_df["Precision"].round(4)
display_df["Recall"] = display_df["Recall"].round(4)
display_df["F1 Score"] = display_df["F1 Score"].round(4)
display_df["Training Time (sec)"] = \
    display_df["Training Time (sec)"].round(2)

print("\nFormatted Results:")

display(display_df)


# 14. TRAINING ACCURACY COMPARISON

plt.figure(figsize=(12, 6))

for kind in ["RNN", "LSTM", "GRU"]:

    plt.plot(
        histories[kind]["accuracy"],
        marker="o",
        label=kind
    )

plt.title("Training Accuracy Comparison")

plt.xlabel("Epoch")

plt.ylabel("Accuracy")

plt.legend()

plt.grid(True)

plt.show()


# 15. VALIDATION ACCURACY COMPARISON

plt.figure(figsize=(12, 6))

for kind in ["RNN", "LSTM", "GRU"]:

    plt.plot(
        histories[kind]["val_accuracy"],
        marker="o",
        label=kind
    )

plt.title("Validation Accuracy Comparison")

plt.xlabel("Epoch")

plt.ylabel("Validation Accuracy")

plt.legend()

plt.grid(True)

plt.show()

# 16. TRAINING LOSS COMPARISON

plt.figure(figsize=(12, 6))

for kind in ["RNN", "LSTM", "GRU"]:

    plt.plot(
        histories[kind]["loss"],
        marker="o",
        label=kind
    )

plt.title("Training Loss Comparison")

plt.xlabel("Epoch")

plt.ylabel("Loss")

plt.legend()

plt.grid(True)

plt.show()


# 17. VALIDATION LOSS COMPARISON

plt.figure(figsize=(12, 6))

for kind in ["RNN", "LSTM", "GRU"]:

    plt.plot(
        histories[kind]["val_loss"],
        marker="o",
        label=kind
    )

plt.title("Validation Loss Comparison")

plt.xlabel("Epoch")

plt.ylabel("Validation Loss")

plt.legend()

plt.grid(True)

plt.show()


# 18. METRIC COMPARISON

metrics = [
    "Accuracy",
    "Precision",
    "Recall",
    "F1 Score"
]

plot_df = df.set_index("Model")[metrics]

plot_df.plot(
    kind="bar",
    figsize=(12, 6)
)

plt.title("RNN vs LSTM vs GRU - Performance Comparison")

plt.xlabel("Model")

plt.ylabel("Score")

plt.ylim(0, 1)

plt.xticks(rotation=0)

plt.grid(axis="y")

plt.legend()

plt.show()


# 19. INDIVIDUAL METRIC BAR CHARTS

for metric in metrics:

    plt.figure(figsize=(8, 5))

    plt.bar(
        df["Model"],
        df[metric]
    )

    plt.title(
        metric + " Comparison"
    )

    plt.xlabel("Model")

    plt.ylabel(metric)

    plt.ylim(0, 1)

    plt.grid(
        axis="y"
    )

    plt.show()


# 20. TRAINING TIME COMPARISON

plt.figure(figsize=(8, 5))

plt.bar(
    df["Model"],
    df["Training Time (sec)"]
)

plt.title("Training Time Comparison")

plt.xlabel("Model")

plt.ylabel("Training Time (seconds)")

plt.grid(
    axis="y"
)

plt.show()


# 21. PARAMETER COUNT COMPARISON

plt.figure(figsize=(8, 5))

plt.bar(
    df["Model"],
    df["Parameters"]
)

plt.title("Model Parameter Comparison")

plt.xlabel("Model")

plt.ylabel("Number of Parameters")

plt.grid(
    axis="y"
)

plt.show()


# 22. ACTUAL VS PREDICTED CLASS DISTRIBUTION

actual_counts = np.bincount(
    y_test,
    minlength=NUM_CLASSES
)

plt.figure(figsize=(14, 5))

plt.plot(
    range(NUM_CLASSES),
    actual_counts,
    marker="o",
    label="Actual"
)

for kind in ["RNN", "LSTM", "GRU"]:

    predicted_counts = np.bincount(
        predictions[kind],
        minlength=NUM_CLASSES
    )

    plt.plot(
        range(NUM_CLASSES),
        predicted_counts,
        marker=".",
        label=kind
    )

plt.title(
    "Actual vs Predicted Class Distribution"
)

plt.xlabel("Class")

plt.ylabel("Number of Samples")

plt.legend()

plt.grid(True)

plt.show()


# 23. FIND BEST MODEL FOR EACH METRIC

print("\n")
print("BEST PERFORMANCE BY METRIC")

for metric in metrics:

    best_index = df[metric].idxmax()

    best_model = df.loc[
        best_index,
        "Model"
    ]

    best_score = df.loc[
        best_index,
        metric
    ]

    print(
        f"{metric}: {best_model} "
        f"({best_score:.4f})"
    )


# 24. FREE MEMORY

gc.collect()

```
